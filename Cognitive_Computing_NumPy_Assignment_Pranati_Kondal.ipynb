```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print(corrected_temperature)

fahrenheit = corrected_temperature * 9 / 5 + 32
print(fahrenheit)

above_32 = corrected_temperature[corrected_temperature > 32]
print(above_32)

count = np.sum(corrected_temperature > 32)
print(count)

print("Vectorization and Boolean indexing process multiple values at once, making operations shorter and faster than using loops.")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(np.sum(steps))
print(np.mean(steps))
print(np.max(steps))
print(np.min(steps))
print(np.sum(steps, axis=0))
print(np.sum(steps, axis=1))

position = np.unravel_index(np.argmax(steps), steps.shape)
print(position)
print(position[0] + 1, position[1] + 1)
```

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:4]
subset[0] = 999

print(original)
print(subset)

copied = original[1:4].copy()
copied[0] = 500

print(original)
print(copied)

matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

print(matrix[0])
print(matrix[-1])
print(matrix[:, 1])
print(matrix[1:3, 2:4])

flat1 = matrix.flatten()
flat2 = matrix.ravel()

print(flat1)
print(flat2)

flat2[0] = 100
print(matrix)
print(flat2)

flat1[1] = 200
print(matrix)
print(flat1)

print(matrix.shape)
print(matrix.ndim)
print(matrix.size)
print(matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

y = np.array([40, 65, 30, 85], dtype=float)

print(X.shape)
print(X.ndim)

print(X.T)
print(X.T @ X)

inverse = np.linalg.inv(X.T @ X)
print(inverse)

beta = inverse @ X.T @ y
print(beta)

new_user = np.array([5, 40, 7], dtype=float)
prediction = new_user @ beta
print(prediction)
```
