```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
m = counts.max()
print(values[counts == m])
for value in values[counts == m]:
    print(value, np.where(x == value)[0])

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
m = counts.max()
print(values[counts == m])
for value in values[counts == m]:
    print(value, np.where(y == value)[0])
```

```python
import numpy as np

pranati = np.linspace(10, 100, 25)

print(pranati)
print(pranati.ndim)
print(pranati.shape)
print(pranati.size)
print(pranati.dtype)
print(pranati.nbytes)

reshaped_pranati = pranati.reshape(1, 25)
transposed_pranati = reshaped_pranati.reshape(25, 1)
print(transposed_pranati)

print(pranati.T)
```

```python
import numpy as np

ucs420_pranati = np.array([[10, 20, 30, 40],
                           [50, 60, 70, 80],
                           [90, 15, 20, 35]])

print(ucs420_pranati)
print(np.mean(ucs420_pranati))
print(np.median(ucs420_pranati))
print(np.max(ucs420_pranati))
print(np.min(ucs420_pranati))

reshaped_ucs420_pranati = ucs420_pranati.reshape(4, 3)
print(reshaped_ucs420_pranati)

resized_ucs420_pranati = np.resize(ucs420_pranati, (2, 3))
print(resized_ucs420_pranati)
```
